## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from sklearn.metrics import f1_score, precision_score, recall_score, accuracy_score

# Actual labels
y_true = [1, 0, 1, 1, 0, 1, 0, 0]

# Model predictions
y_pred = [1, 0, 1, 0, 0, 1, 0, 1]

# Calculate metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)

# Success criteria
if f1 >= 0.85:
    print("Model Performance: Good ✅")
else:
    print("Model Performance: Needs Improvement ⚠️")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Fixed rule-based approach

def rule_based_prediction(url):
    suspicious_words = ["free", "prize", "verify", "login", "urgent"]
    
    for word in suspicious_words:
        if word in url.lower():
            return 1   # Scam
    return 0           # Legitimate


# Example URLs
urls = [
    "free-prize-login.com",
    "google.com",
    "verify-account-security.com",
    "amazon.com"
]

# Apply fixed rule
for url in urls:
    print(url, "->", rule_based_prediction(url))
```
